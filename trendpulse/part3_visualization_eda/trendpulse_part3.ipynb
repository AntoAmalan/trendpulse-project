```python
# Part 3: Visualization, EDA & Business Thinking
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns # type: ignore
import plotly.express as px

df = pd.read_csv('data/USvideos.csv')

# 1. Matplotlib Visualizations
plt.figure(figsize=(10, 5))
sns.histplot(np.log1p(df['views']), bins=50, kde=True)
plt.title("Log-Scaled Distribution of Video Views")
plt.xlabel("Log(Views + 1)")
plt.ylabel("Frequency")
plt.show()

# 2. Plotly Interactive Chart
fig = px.scatter(df.sample(1000), x="views", y="likes", color="category_id",
                 hover_data=["title", "channel_title"], log_x=True, log_y=True,
                 title="Interactive Views vs. Likes by Category")
fig.show()

# 3. Funnel Analysis Simulation
total_videos = df['video_id'].nunique()
median_views = df['views'].median()
above_median = df[df['views'] > median_views]['video_id'].nunique()
high_comments = df[df['comment_count'] > 1000]['video_id'].nunique()

print(f"Funnel Analysis:\n- Total Unique Videos: {total_videos}\n- Above Median Views: {above_median}\n- High Comments (>1k): {high_comments}")
```
